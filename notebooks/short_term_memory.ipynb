```python
from collections import deque

class ShortTermMemory:
    def __init__(self, max_messages=5):
        self.buffer = deque(maxlen=max_messages)

    def add(self, role, message):
        self.buffer.append({"role": role, "message": message})

    def get_context(self):
        return list(self.buffer)

    def __str__(self):
        return "\n".join([f"{m['role']}: {m['message']}" for m in self.buffer])
```

```python
memory = ShortTermMemory(max_messages=5)
```

```python
memory.add("user", "안녕!")
memory.add("assistant", "안녕하세요?")
memory.add("user", "날씨 어때?")
memory.add("assistant", "맑아요.")
memory.add("user", "점심 뭐 먹지?")
memory.add("assistant", "라면 추천드려요.")
memory.add("user", "저녁은 뭘 먹지?")
memory.add("assistant", "글쎄..")
memory.add("assistant", "뭐가 좋을까?")
```
